```python
import json
from pathlib import Path

parents = [json.loads(l) for l in Path("../data/processed/chunks_parents.jsonl").read_text().splitlines()]
children = [json.loads(l) for l in Path("../data/processed/chunks_children.jsonl").read_text().splitlines()]

# Find a German DSGVO child and trace it back to its parent
de_children = [c for c in children if c["metadata"]["language"] == "de"]
sample = de_children[20]

print(f"CHILD chunk ({sample['token_count']} tokens):")
print(f"  ID: {sample['chunk_id']}")
print(f"  parent_id: {sample['parent_id']}")
print(f"  Section: {sample['metadata']['section_heading']}")
print(f"\n  Text for embedding (with prefix):")
print(f"  {sample['text_for_embedding'][:400]}...\n")

# Find the parent
parent = next(p for p in parents if p["chunk_id"] == sample["parent_id"])
print(f"PARENT ({parent['token_count']} tokens, {len(parent['child_ids'])} children):")
print(f"  Pages: {parent['metadata']['page_start']}-{parent['metadata']['page_end']}")
print(f"  Body: {parent['text'][:400]}...")
```
